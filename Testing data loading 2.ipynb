```python
import nibabel as nib

# Load the image
abc_img = nib.load(path_to_images+ abc + '001_adc.nii.gz')

# Get image data as a NumPy array
abc_data = abc_img.get_fdata()

# Print shape
print(abc_data.shape)
```

```python
import os
import re

folders = {
    "im_adc": "images/adc_to_t2w",
    "im_hbv": "images/hbv_to_t2w",
    "im_t2w_ax": "images/t2w_ax_to_t2w",
    "lab_intersections": "labels/intersection/adc_to_t2w",
    "majority": "labels/majority/adc_to_t2w" 

}

pattern = re.compile(r"(\d+)_")  # matches xxx_

def extract_ids(folder):
    ids = set()
    for f in os.listdir(folder):
        match = pattern.search(f)
        if match:
            ids.add(int(match.group(1)))
    return ids

# Extract IDs from each folder
ids_per_folder = {
    name: extract_ids(path_to_data+path)
    for name, path in folders.items()
}

# ---- Compare folders ----
all_ids = set.union(*ids_per_folder.values())

print("Missing files per folder:")
for name, ids in ids_per_folder.items():
    missing = sorted(all_ids - ids)
    print(f"{name}: {missing}")

# ---- Check numeric gaps ----
sorted_ids = sorted(all_ids)
gaps = [
    i for i in range(sorted_ids[0], sorted_ids[-1] + 1)
    if i not in all_ids
]

print("\nNumeric gaps:")
print(gaps)
```
