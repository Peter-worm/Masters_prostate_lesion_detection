```python
import pandas as pd
import io

import pandas as pd
import re

def extract_fold_and_clean_columns(df):
    fold_value = None
    new_columns = {}

    # regex: fold_2 patient cl_1 : AUC
    fold_regex = r"^fold_(\d+)\s+(.*)$"

    for col in df.columns:
        m = re.match(fold_regex, col)
        if m:
            fold_value = int(m.group(1))  # extract fold number
            new_columns[col] = m.group(2)  # remove the "fold_2 " part
        else:
            new_columns[col] = col  # keep unchanged

    # apply cleaned column names
    df = df.rename(columns=new_columns)

    # insert fold column
    df["Fold"] = fold_value

    return df

def load_multiheader_csv(path):
    with open(path, "r") as f:
        content = f.read().strip()

    # split blocks by blank lines
    blocks = [b.strip() for b in content.split("\n\n") if b.strip()]

    dfs = []
    for block in blocks:
        # each block has exactly 2 lines: header + data
        try:
            df = pd.read_csv(io.StringIO(block))
            df = extract_fold_and_clean_columns(df)
            print(df)
            dfs.append(df)
        except Exception as e:
            print("Error parsing block:\n", block)
            raise e

    # combine all parsed blocks
    final_df = pd.concat(dfs, ignore_index=True)
    return final_df


# Usage
df = load_multiheader_csv(path)
# print(df)


# Example usage after loading the file:
df = load_multiheader_csv("your_file.txt")

# Apply cleaning to each row
cleaned_rows = []
for _, row in df.iterrows():
    cleaned = extract_fold_and_clean_columns(pd.DataFrame([row])).iloc[0]
    cleaned_rows.append(cleaned)

df_clean = pd.DataFrame(cleaned_rows)

print(df_clean.head())
```
